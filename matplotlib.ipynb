```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5,6]
y = [10,60,31,40,50,67]
plt.scatter(x,y ,marker="s",color = "red",s=100,alpha=0.7,
            edgecolors="black",linewidths=1.5,label = "students")
plt.title("scatter chat")
plt.xlabel("x axis")
plt.ylabel("y axis")

for i in range(len(x)):
    plt.text(x[i],y[i],str(y[i]))
plt.legend()    
    
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
x = [1,2,3,4,5,6]
y = [10,60,31,40,50,67]
plt.scatter(x,y ,marker="s",color = "red",s=100,alpha=0.7,
            edgecolors="black",linewidths=1.5,label = "students")
plt.title("scatter chat")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.legend(labelcolor = "blue")   

for i in range(len(x)):
    plt.text(x[i],y[i],str(y[i]))
legend_marker = Line2D(
    [0], [0],
    marker="s",
    markerfacecolor="black",
    markeredgecolor="black",
    color="black",
    linestyle="None",
    markersize=8,
    label="students"
)   
  
plt.legend(handles=[legend_marker],labelcolor="blue")
 
plt.grid()
plt.show()
```
