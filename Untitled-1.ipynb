A loop is used to execute the same block of code repeatedly.

Definition

A for loop is used when we want to repeat something for each item in a sequence or for a known range of values.
Syntax
for variable in sequence:
    statement

```python
# Palindrome Number

# A palindrome reads the same forward and backward.

n = 11232
original = n
reverse = 0
while n >0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10
if original == reverse :
    print ("palindrome")
else:
    print("not a palindrome")        
    
#  same expale   
n = 121
original = n
reverse = 0
while n >0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10
if original == reverse :
    print ("palindrome")
else:
    print("not a palindrome")  
```

```python
# Nested While Loop
# A loop inside another loop is called a nested loop.
i = 1
while i <= 3:
    j = 1
    while j  <= 3:
        print(i,j)
        j = j +1
    i = i +1     
```

```python
# 1. Find duplicate elements without using count()
# Given a list, find all duplicate elements using Set.
a = [10,20,30,20,40,10,50,30]
seen = set()
duplicates = set()
for i in a:
    if i in seen:
        duplicates.add(i)
    else:
        seen.add(i)
print("duplicates: ",duplicates)        
            
```

```python
# 2. Find elements that occur exactly once
# Find values that appear only once.
a = [10,20,30,40,50,10,50,20]
unique = set()
for i in a:
    if a.count(i) == 1:
        unique.add(i)
print(unique)        
```

```python
# Better interview approach
a = [10,20,30,40,50,60,10]
seen = set()
duplicates = set()
for i in a:
    if i in seen:
        duplicates.add(i)
    else:
        seen.add(i)
result = seen - duplicates
print(result)            
```

```python
# 3. Find common elements in three List
a = [10,20,30,40]
b = [20,30,50]
c = [30,20,60]
result = set(a) & set(b) & set(c)
print(result)
```

```python
# Find elements common to exactly two Sets
a = {10,20,30}
b = {20,30,40}
c = {30,40,50}
all_values = a | b | c
result = {
    x for x in all_values
    if (x in a) + ( x in b) + (x in c)==2
}
print(result)
```

```python
# Find missing numbers from a sequence
a = {1,2,3,4,5,6,7,10}
expected = set(range(1,11))
missing = expected - a
print(missing)
```

```python
#  Find duplicate words in a sentence
sentence = "python sql python excl sql python"
words = sentence.split()
seen = set()
duplicates = set()
for word in words:
    if word in seen:
        duplicates.add(word)
    else:
            seen.add(word)
print(duplicates)            
            
```

```python
# Check whether two Lists contain the same unique values
a = [10,20,30,20]
b = [30,10,20]
if set(a) == set(b):
    print("same unique values")
else:
    print("different values")    
    
    
```

```python
# Check whether two Lists are completely disjoint
a = [10,20,30]
b = [40,50,60]
if set(a).isdisjoint(set(b)):
    print("no common elementes")
else:
    print("common elements exist")    
```

```python
# Find the largest and second-largest values using Set
a = {10,50,20,80,30}
largest = max(a)
a.remove(largest)
second_largest = max(a)
print("largest",largest)
print("second largest: ",second_largest)
```

```python
# Check whether one List is a subset of another
a = [10,20]
b = [10,20,30,40]
if set(a).issubset(set(b)):
    print("a is subset if b")
else:
    print("a is not subset of b")    
```
