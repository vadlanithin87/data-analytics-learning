```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ravi", "Sita", "Arjun", "Priya", "Ravi", "Kiran", "Anu", "Sita", "Rahul", None],
    "Age": [22, 24, 21, 25, 22, 23, None, 24, 26, 25],
    "City": ["Hyderabad", "Kurnool", "Hyderabad", "Vijayawada", "Hyderabad",
             "Kurnool", "Guntur", "Kurnool", "Guntur", None],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Male", "Female", "Female", "Male", "Female"],
    "Marks": [85, 92, 76, 88, 85, None, 95, 92, 67, 80],
    "Salary": [30000, 35000, 28000, 40000, 30000, 32000, 45000, 35000, 27000, 31000]
})

print(df)
df
```
