```python
# 4. Movie Ticket 🎬

# First decide ticket price using age:

# Below 12 → ₹120
# 12–59 → ₹200
# 60+ → ₹140

# Then add:

# Weekend → +₹30
# 3D → +₹70
# Both → ₹20 discount on extra charges

age = int(input("Enter age: "))
weekend = input("Weekend? yes/no: ")
three_d = input("3D? yes/no: ")

if age < 12:
    price = 120
elif age <= 59:
    price = 200
else:
    price = 140

if weekend == "yes":
    price += 30

if three_d == "yes":
    price += 70

if weekend == "yes" and three_d == "yes":
    price -= 20

print("Ticket Price:", price)
```

```python
# 5. Internet Data Plan 📱

# First check how much data was used:

# ≤ 2 GB → ₹199
# ≤ 5 GB → ₹399
# ≤ 10 GB → ₹599
# > 10 GB → ₹599 + ₹50 for each extra GB

# 5G + usage above 5 GB → ₹100 discou

gb = float(input("enter gb : "))
five_g = input("5g? yes/no")
if gb <=2:
    charge = 199
elif gb <= 5:
    charge = 399
elif gb <=10:
    charge = 599
else:
    extra = gb - 10
    charge = 599 + extra * 50
if five_g == "yes" and gb >5:
    charge -= 100
print("charge: ",charge)                
    
        
```

```python
# 6. Flight Baggage ✈️

# First find free baggage allowance:

# Economy → 15 kg
# Business → 30 kg
# First → 40 kg

# If extra baggage exists:

# First 5 kg → ₹500/kg
# Next 10 kg → ₹700/kg
# # Remaining → ₹1000/kg

ticket = input("ticket class: ").lower()
weight = float(input("baggage weight: "))
if ticket == "economy":
    allowance = 15
elif ticket == "business" :
    allowance = 30
else:
    allowance = 40
extra = weight -allowance

if extra <=0:
    charge =0
elif extra <=5:
    charge = extra * 500
elif extra <=15:
    charge = 5 * 500 + (extra - 5) * 700
else:
    charge = 5 * 500 + 10 * 700 + (extra - 15) * 1000
print("charge",charge)   
                        
```

```python
# 7. Online Course Discount 📚

# First check course price.

# Then check student/professional/other.

# Finally check whether previous courses are more than 5.

price = float(input("Course price: "))
person = input("Student/Professional/Other: ").lower()
courses = int(input("Previous courses: "))

discount = 0

if price >= 10000:

    if person == "student":
        discount = 25
    elif person == "professional":
        discount = 15
    else:
        discount = 10

elif price >= 5000:

    if person == "student":
        discount = 15
    elif person == "professional":
        discount = 10
    else:
        discount = 5

final_price = price - (price * discount / 100)

if courses > 5:
    final_price -= 500

print("Final Price:", final_price)
```

```python
# 8. Cab Fare 🚕

# First calculate distance charge.

# Base = ₹80

# First 5 km → ₹12/km
# Next 10 km → ₹15/km
# Above 15 km → ₹20/km

# Then:

# Night → +₹100
# Distance >20 → -₹50
# Premium → no night charge

km = float(input("Enter km: "))
night = input("Night? yes/no: ")
premium = input("Premium? yes/no: ")

fare = 80

if km <= 5:
    fare += km * 12

elif km <= 15:
    fare += 5 * 12
    fare += (km - 5) * 15

else:
    fare += 5 * 12
    fare += 10 * 15
    fare += (km - 15) * 20

if night == "yes" and premium != "yes":
    fare += 100

if km > 20:
    fare -= 50

print("Total Fare:", fare)
```

```python
# 9. Warehouse Storage 📦

# First check weight:

# ≤10 kg → ₹100
# ≤25 kg → ₹200
# ≤50 kg → ₹400
# >50 kg → ₹700

# Then storage days and fragile status can add charges.

# Easy code
weight = float(input("Weight: "))
days = int(input("Days: "))
fragile = input("Fragile? yes/no: ")

if weight <= 10:
    charge = 100
elif weight <= 25:
    charge = 200
elif weight <= 50:
    charge = 400
else:
    charge = 700

if days <= 3:
    extra = 0
elif days <= 7:
    extra = (days - 3) * 50
else:
    extra = 4 * 50 + (days - 7) * 100

charge += extra

if fragile == "yes":
    charge += 150

print("Total Charge:", charge)
```

```python
# 10. Employee Attendance 👨‍💼

# This one looks difficult, but remember only two checks:

# First → Attendance
# 95+      → Excellent group
# 80–94    → Normal group
# Below 80 → Warning group
# Second → Performance

# Then performance decides the exact message.

# Finally:

# Attendance <75%
# → Salary Deduction Applicable


attendance = float(input("Attendance: "))
performance = float(input("Performance: "))

if attendance >= 95:

    if performance >= 90:
        status = "Excellent"

    elif performance >= 75:
        status = "Good"

    else:
        status = "Attendance Excellent, Performance Needs Improvement"


elif attendance >= 80:

    if performance >= 90:
        status = "Performance Excellent"

    else:
        status = "Normal"


else:

    if performance >= 90:
        status = "High Performance but Attendance Warning"

    else:
        status = "Attendance Warning"


print("Status:", status)

if attendance < 75:
    print("Salary Deduction Applicable")

```

A basic while loop has four important parts:

while i <= 5:      # 2. Condition

Problem Statement
Design and develop a Inventory Management System using Python that manages
stock for multiple products in a shop.
The system should initially store the available quantity for different products such
as Rice, Sugar, and Oil. The program must allow the user to repeatedly perform
operations on these products using a menu.
The system should provide the following functionalities:
1. Product Selection
o Allow the user to select a product (Rice, Sugar, or Oil).
2. Stock Operations
Based on the selected product, the user should be able to:
o Add Stock: Increase the stock quantity of the selected product.
o Sell Stock: Decrease the stock quantity if sufficient stock is available.
o Check Stock: Display the current available stock of the selected
product.
3. Validation
o If the user tries to sell more quantity than available, display an
appropriate message.
o Handle invalid menu choices properly.
4. Program Control
o The program should allow multiple operations using a loop.
o The program should terminate only when the user chooses the exit
option.

Constraints / Requirements
 Use if–elif–else statements for decisi
 on making.
 Use a for loop to allow multiple operations.
 Use only basic arithmetic operations.

```python
# Initial stock
rice = 100
sugar = 80
oil = 50

# Perform 5 operations
for i in range(5):

    print("\n--- Inventory Management System ---")
    print("1. Rice")
    print("2. Sugar")
    print("3. Oil")
    print("4. Exit")

    product = int(input("Select Product: "))

    if product == 4:
        print("Program ended.")
        break

    elif product == 1:
        stock = rice
        product_name = "Rice"

    elif product == 2:
        stock = sugar
        product_name = "Sugar"

    elif product == 3:
        stock = oil
        product_name = "Oil"

    else:
        print("Invalid product choice.")
        continue

    print("\nSelected Product:", product_name)
    print("1. Add Stock")
    print("2. Sell Stock")
    print("3. Check Stock")

    operation = int(input("Select Operation: "))

    if operation == 1:
        quantity = int(input("Enter quantity to add: "))
        stock = stock + quantity

        if product == 1:
            rice = stock
        elif product == 2:
            sugar = stock
        else:
            oil = stock

        print("Stock added successfully.")
        print("Current Stock:", stock)

    elif operation == 2:
        quantity = int(input("Enter quantity to sell: "))

        if quantity <= stock:
            stock = stock - quantity

            if product == 1:
                rice = stock
            elif product == 2:
                sugar = stock
            else:
                oil = stock

            print("Stock sold successfully.")
            print("Remaining Stock:", stock)

        else:
            print("Insufficient stock.")
            print("Available Stock:", stock)

    elif operation == 3:
        print("Current Stock:", stock)

    else:
        print("Invalid operation choice.")

else:
    print("\nMaximum operations completed.")
```

```python
# creating a dicionary
student = {
    "name": "nithin",
    "marks": 90,
    "age": 22,
    "course": "DA"
}
print(student)
print(student.keys())
print(student.values())
print(student.items())
```

```python
# Dictionary is Mutable
# Definition

# Mutable means we can change the dictionary after creating it.
student = {
    "name": "nithin",
    "marks": 95
    }
student["marks"]= 90
print(student)
```

```python
# Dictionary is Ordered

# Modern Python dictionaries maintain insertion order.
student = {
    "name": "nithin",
    "marks":85,
    "age": 22
}
print(student)
```

```python
# Dictionary Does NOT Allow Duplicate Keys
# The second value replaces the first value.
# Important
# Duplicate values → allowed
# Duplicate keys   → not maintained

student = {
    "name":"nithin",
    "name":"rahul"
}
print(student)
```

```python
# Accessing Dictionary Values
# Use the key.
student = {
    "name": "nithin",
    "age": 22,
    "marks": 85
}
print(student["name"])
print(student["marks"])
```

```python
# Accessing with get()
student = {
    "name": "nithin",
    "age": 22
}
print(student.get("name"))
```

```python
# Adding a New Item
student = {
    "name": "nithin",
    "age": 22
}
student["marks"] = 85
print(student)
```

```python
# Updating an Existing Value
student = {
    "name":"nithin",
    "marks": 80
}
student ["marks"]= 95
print(student)
```

```python
# update()
# Used to add or update multiple key-value pairs.
student = {
    "name": "nithin",
    "age": "22"
}
student.update({
    "marks": 85,
    "city": "kurnool"
})
print(student)
```

```python
# . pop()
# Removes an item using its key.
student = {
    "name": "nithin",
    "marks": 85,
    "age": 22
}
print(student)
student .pop("age")
print(student)
```

```python
# popitem()
# Removes the last inserted key-value pair.
student = {
    "name": "nithin",
    "age": 22,
    "marks": 85
}
student.popitem()
print(student)
```

```python
# del
# Deletes an item.
student = {
    "name": "nithin",
    "age": 22
}
del student["age"]
print(student)
```

```python
# clear()  --> Removes all items.
student = {
    "name":"nithin",
    "age": 22
}
student.clear()
print(student)
```

```python
# len()
# Returns the number of key-value pairs.
student = {
    "name": "Nithin",
    "age": 22,
    "marks": 85
}

print(len(student))
```

```python
# Frequency Counting ⭐⭐⭐

# This is one of the most important dictionary interview programs.
numbers = [1,2,2,3,3,3,4]
freq = {}
for i in numbers:
    if i in freq:
        freq[i] +=1
    else:
        freq[1]= 1
print(freq)            
```

```python
#  characrter frequency
word = "banana"
freq = {}
for ch in word:
    freq [ch] = freq.get(ch,0) +1
print(freq)
```

```python
#  find duplicate vlaues
numbers = [1,2,2,3,4,4,5]
freq= {}
for i in numbers:
    freq[i] = freq.get(i,0)+1
for key ,value in freq.items():
    if value >1:
        print(key)    
```
