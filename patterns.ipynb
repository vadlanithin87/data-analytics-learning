```python
#  square patterns
for i in range(4):
    for j in range(4):
        print("*",end=" ")
    print()    
```

```python
#  incresing triangle
for i in range (1,5):
    for  j in range(i):
        print('*',end=" ")
    print()


#decresing trainge
for i in range(4,0,-1):
    for j in  range(i):
        print("*",end=" ")
    print()     
    
    
#  numbers pattern    #
for i in range (1,5):
    for j in range(1,i+1):
        print(j,end=" ")
    print()    
    
    
# same number pattern 
for i in range(1,5):
    for j in  range(i):
        print(i,end=" ")
    print()         
```

```python
#  increasing number pattern
for i in range(1,5):
    for j in range(1,i+1):
        print(j,end=" ")
    print()    
    
    
    
# reverse number pattern
for i in range (4,0,-1):
    for j in range(4,4-i,-1):
        print(j,end= " ")
    print()        
        
```

```python
#  continuous number pattern
x = 1
for i in range(1,5):
    for j in range(i):
        print(x,end=" ")
        x += 1
    print()    
    
    
# star right traingle
for i in range(1,6):
    for j in range(i):
        print("*",end=" ")
    print()       
    
# reversed trainge
for i in range(5,0,-1):
    for j in range(i):
        print("*",end=" ")
    print()   
         
```

```python
#  right aligned traingle
for i in range(1,6):
    for j in range(5,-i):
        print(" ",end="")
    
    for j in range(i):
        print("*",end="")    
    print()
```

```python
#  pyramid
for  i in range(1,6):
    for j in range(5,-i):
        print(" ",end=" ")
    for j in range(2*i-1):
        print("*",end=" ")
    print()        
```

```python
# daimond star pattern
for i in range(1,5):
    for j in range(4-i):
        print("",end="")
    for j in range(2*i-1):
        print("*",end="")
    print()
for i in range(3,0,-1):
    for j in range(4-i):
        print(" ",end="")
    for j in range(2 * i -1):
        print("*",end="")
    print()                    
```
