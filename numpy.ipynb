What is ndarray?

English:
ndarray means N-dimensional array. NumPy lo arrays ni represent cheyyadaniki main data structure.





```python
import numpy  as np
arr = np.array([10,20,30,40])
print(arr)
print(type(arr))
```

4. 1D / 2D / 3D Arrays
1D Array



```python
arr1 = np.array([10,20,30,40])
print(arr1)
```

2D Array

Rows + columns.

```python
arr2 = np.array([
    [10,20],
    [30,40]
])
print(arr2)
```

5. Array Creation
Most common method:
np.zeros()
np.ones()
np.arange()
np.linspace()
np.full()

```python
arr = np.array([1,2,3,4])
print(arr)
```

6. np.zeros()

English: Creates an array filled with zeros.



```python
arr = np.zeros(5)
print(arr)
```

7. np.ones()

English: Creates array filled with 1.

```python
arr = np.ones(5)
print(arr)
```

8. np.arange()

English: Creates numbers within a range.



```python
arr = np.arange(1,10)
print(arr)

arr = np.arange(1, 11, 2)
print(arr)
```

9. np.linspace()

English: Creates a specified number of equally spaced values.

```python
arr = np.linspace(0,10,5)
print(arr)
```

10. np.full()

English: Creates array with the same specified value.

```python
arr = np.full(5,10)
print(arr)

arr = np.full((2, 3), 7)
print(arr)
```

11. ndim

English: Number of dimensions.


```python
arr = np.array([[10,20],[30,40]])
print(arr.ndim)
```

12. shape

English: Tells number of rows and columns/dimensions

```python
arr = np.array([
    [10,20,30],
    [40,50,60]
])
print(arr.shape)
```

13. size

English: Total number of elements.

```python
arr = np.array([[10,20,30],
              [40,50,60]
])        
print(arr.size)
```

14. dtype

English: Data type of array elements.

int
float
bool
complex
string

```python
arr = np.array([10,20,30])
print(arr.dtype)

arr = np.array([10.5,20.5])
print(arr.dtype)

arr = np.array([10==10,10<5])
print(arr.dtype)
```

15. astype()

English: Converts array data type.


```python
arr = np.array([1,2,3,4])
new_arr = arr.astype(float)
print(new_arr)
print(new_arr.dtype)
```

```python
# Float → int
arr = np.array([10.5,20.8,30.2])
new_arr = arr.astype(int)
print(new_arr)
```

```python
import numpy as np
arr = np.array ([10,20,30,40])
print(arr[0])
print(arr[2])
```

```python
# 2D Indexing
arr= np.array([[10,20],
               [30,40]])
print(arr[0,1])
```

17. Slicing 

Slicing = array lo portion select 

```python
arr = np.array([10,20,30,40,50])
print(arr[1:4])
print(arr[:3])
print(arr[2:])
print(arr[::2])
```

18. Reshaping 

Reshape = array shape change 

```python
import numpy as np
arr = np.array([1,2,3,4,5,6])
print(6)
```

```python
new_arr = arr.reshape(2,3)
print(new_arr)
```

19. flatten()

English: Converts multidimensional array into 1D.

```python
arr = np.array([[1,2],[3,4]])
nwe_arr = arr.flatten()
print(new_arr)
```
