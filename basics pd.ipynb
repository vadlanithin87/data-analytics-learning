```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Kiran", "Anu"],
    "Age": [21, 22, 20, 23],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad"],
    "Marks": [85, 92, 78, 88],
    "percent":[85.0,92.0,78.0,88.0]
}

df = pd.DataFrame(data)


print(df)
df
```

```python
# 10. Find students with marks between 80 and 90
# Important: Pandas uses & for AND and | for OR.
print(df[(df["Marks"] >= 80) & (df["Marks"] <= 90)])
```

```python
# 12. Find highest marks
print(df["Marks"].max())
#  13 --> find the  lowest marks
print(df["Marks"].min())
#  14. find the average marsk
print(df["Marks"].mean())
#  15. find the total marks
print(df["Marks"].sum())
# find the total  percent
print(df["percent"].sum())
```

```python
# 16. Add a new column
# Question: Add a Result column.
df["result"] = df["Marks"] >=40
df

#  addd % column
df ["%"] = df["percent"] >=40
df
```

```python
# # 17. Create Pass/Fail column
df["result"] = df["Marks"].apply(
    lambda x: "pass" if x >= 40 else "fail"
)
print(df)

df["%"] = df["percent"].apply(
    lambda x: "pass" if x >=95 else "fail"
)
# print(df)
df
```

```python
#  18 create a grade column
df["grade"] = df['Marks'].apply(
    lambda x : "A" if x >=90
    else "B" if x >=80
    else "C"
)
# print(df)
df
```

```python
# 24. Group by City
# Question: Find average marks for each city.
print(df.groupby("City")["Marks"].mean())
print(df.groupby("City")["Marks"].max())

```
