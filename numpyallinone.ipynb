```python
#  iqr interquartile range
a = np.array([10,12,15,18,20,22,25,100])

q1 = np.percentile(a,25)
q3 = np.percentile(a,75)
iqr = q3-q1
lower  = q1 - 1.5 * iqr
upper = q3 + q3 +iqr

outliers = a[(a< lower) | (a> upper)]
print("q1: ",q1)
print("q3: ",q3)
print("iqr: ",iqr)
print("lower: ",lower)
print('upper: ',upper)
print("outliers: ",outliers)
```
